```python
"""
BTL 46 - Heart Disease
00_Data_Preparation.py

"""

import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib

# ====================== CAU HINH ======================
RANDOM_STATE = 42
TEST_SIZE = 0.2
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)

print("=" * 60)
print("BTL 46 - HEART DISEASE | DATA PREPARATION")
print("=" * 60)

# ====================== 1. LOAD DU LIEU ======================
# Lay processed Cleveland data tu UCI
column_names = [
    "age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
    "thalach", "exang", "oldpeak", "slope", "ca", "thal", "target"
]

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

df = pd.read_csv(url, header=None, names=column_names, na_values="?")

print("\n[1] Load du lieu thanh cong")
print(f"Shape ban dau: {df.shape}")
print(f"So gia tri thieu:\n{df.isna().sum()}")

# ====================== 2. LAM SACH DU LIEU ======================
# Xoa dong co missing value
df = df.dropna().reset_index(drop=True)

# Chuyen target thanh nhi phan: 0 = khong benh, 1 = co benh
df["target"] = (df["target"] > 0).astype(int)

print(f"\n[2] Sau khi lam sach: {df.shape}")
print(f"Phan bo target:\n{df['target'].value_counts()}")
print(f"Ty le: {df['target'].value_counts(normalize=True).round(3).to_dict()}")

# ====================== 3. EDA NHANH ======================
print("\n[3] Thong tin co ban:")
print(df.info())
print("\nMo ta thong ke:")
print(df.describe().round(2))

corr_with_target = df.corr()["target"].sort_values(ascending=False)
print("\nTuong quan voi target:")
print(corr_with_target)

# ====================== 4. TACH FEATURE - TARGET ======================
X = df.drop("target", axis=1)
y = df["target"]

feature_names = X.columns.tolist()
print(f"\n[4] So dac trung: {len(feature_names)}")
print("Danh sach dac trung:", feature_names)

# ====================== 5. CHIA TRAIN / TEST ======================
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    stratify=y,
    random_state=RANDOM_STATE
)

print(f"\n[5] Chia du lieu (stratify, random_state={RANDOM_STATE})")
print(f"X_train: {X_train.shape} | y_train: {y_train.shape}")
print(f"X_test : {X_test.shape}  | y_test : {y_test.shape}")
print(f"Ty le target trong train: {y_train.value_counts(normalize=True).round(3).to_dict()}")
print(f"Ty le target trong test : {y_test.value_counts(normalize=True).round(3).to_dict()}")

# ====================== 6. CHUAN HOA ======================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\n[6] StandardScaler da fit tren X_train va transform ca train/test")

# ====================== 7. LUU FILE ======================
np.save(os.path.join(DATA_DIR, "X_train.npy"), X_train_scaled)
np.save(os.path.join(DATA_DIR, "X_test.npy"), X_test_scaled)
np.save(os.path.join(DATA_DIR, "y_train.npy"), y_train.values)
np.save(os.path.join(DATA_DIR, "y_test.npy"), y_test.values)
joblib.dump(scaler, os.path.join(DATA_DIR, "scaler.joblib"))

# Luu them ten dac trung va ban raw (chua scale)
np.save(os.path.join(DATA_DIR, "feature_names.npy"), np.array(feature_names))
X_train.to_csv(os.path.join(DATA_DIR, "X_train_raw.csv"), index=False)
X_test.to_csv(os.path.join(DATA_DIR, "X_test_raw.csv"), index=False)

print("\n[7] Da luu cac file vao thu muc data/:")
for f in sorted(os.listdir(DATA_DIR)):
    print(f"  - {f}")

print("\n" + "=" * 60)
print("HOAN TAT DATA PREPARATION")
print("=" * 60)
```
